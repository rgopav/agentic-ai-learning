## LangChain with Gemini

This notebook demonstrates how to use `langchain` with Google's Gemini models to create a simple conversational agent. It covers installing necessary libraries, setting up API authentication, initializing the LLM, and invoking the agent with a user query.

```python
!pip install langchain
!pip install langchain-google-genai
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent

from google.colab import userdata
API_KEY=userdata.get('GEMINI_API_KEY')
```

### Initialize the Chat Model and Agent

Here, we initialize the `ChatGoogleGenerativeAI` model with a specified model and temperature. We then create a generic agent using `create_agent` and invoke it with a user message.

```python
llm = ChatGoogleGenerativeAI(
  model="gemini-2.5-flash-lite",
  temperature=0,
  api_key=API_KEY
)

chat_agent = create_agent(
  model=llm
)

response = chat_agent.invoke(
    {
        "messages":
        [{
            "role": "user",
            "content": "Where are the top 3 popular EV charging networks. Where does IONNA stand?"
        }]
    }
)

print(response["messages"][-1].content)
```

#### Note on Model Name:

The model `gemini-2.5-flash-lite` might be a specific or preview version. For general use and stability, you might consider using publicly available models like `gemini-pro`, `gemini-1.5-flash`, or `gemini-1.5-pro` depending on your needs. You can find the latest available models in the [Gemini API documentation](https://ai.google.dev/models/gemini).
