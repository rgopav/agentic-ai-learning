Automated **Triage Engines** use LLMs to immediately analyze the customer's text, extract the key issues, determine sentiment/urgency, and route the ticket to the correct department (e.g., Billing, Tech Support, Legal).

### 🎯 Your Challenge:
1. **Initialize the Model**: Load your environment and instantiate `ChatGoogleGenerativeAI` using the `gemini-1.5-flash` model. Set `temperature=0` (we want deterministic classifications, not creative fabrications!).
2. **Draft the Input**: We provide a realistic, angry customer support email about a double billing issue and system access crash.
3. **Build the Prompt**: Craft a prompt instructing Gemini to analyze the text and return a clean, structured report in plain text containing:
   - **Sentiment**: (Positive / Negative / Neutral)
   - **Priority**: (High / Medium / Low)
   - **Primary Issue**: A concise one-sentence summary of the user's problem.
   - **Routed Department**: (Billing / Tech Support / Account Management)
4. **Execute**: Send the request and print **only** the clean, classified response text.

```python
# Step 1: Install required packages if you haven't already
# %pip install langchain langchain-google-genai python-dotenv
```

```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage
from langchain_core.prompts import ChatPromptTemplate

# Load environment variables from .env file
load_dotenv('.env')

# Verify API key detection
if not os.getenv("GOOGLE_API_KEY"):
    print("⚠️ Warning: GOOGLE_API_KEY is not set. Make sure it is defined in your .env file in this directory!")
else:
    print("✅ GOOGLE_API_KEY successfully detected!")
```

```python
# 2. Real-world Customer Ticket Input
customer_email = """
Subject: URGENT: Double Charge and System Locked Out!

Hi customer support,
I am writing because I was billed twice ($99.00 x 2) on my invoice #INV-88749 yesterday. 
To make things worse, when I tried to log in to my premium account to check the billing history, 
your portal gave me a '403 Forbidden Access' error! I've cleared my cache and cookies and it's still locked.
I need this double charge refunded immediately and my login access restored. I have a presentation 
with clients tomorrow morning and need this resolved now!

Thanks,
Marcus Vance
"""
```

```python
# TODO: Initialize the Gemini Chat Model for structured classification
# Set model='gemini-1.5-flash' and temperature=0.0

llm = ChatGoogleGenerativeAI(
  model = "gemini-2.5-flash-lite",
  temperature = 0
)
```

```python
# TODO: Analyze the ticket and output a structured triage report
# 1. Write an explicit prompt instructing Gemini to classify 'customer_email'.
#    Instruct it to extract: Sentiment, Priority, Primary Issue, and Routed Department.
# 2. Wrap the prompt inside a HumanMessage object.
# 3. Invoke the model and print the result content.

message = HumanMessage(content=f"""Please classify the following customer email by providing the following.
Sentiment
Priority
Primary Issue 
Routed Department

Customer Email:
{customer_email}
""")

response = llm.invoke([message])
print(response.content)
```
