### 💼 Business Case & Real-World Context:
Personalized outreach is the cornerstone of B2B sales. However, sales representatives waste hours writing custom emails. Companies build **Dynamic Copywriter Engines** that combine CRM client data (prospect name, pain points, company metrics) with highly refined prompt templates to automatically generate custom, high-converting outreach emails.

### 💡 Portfolio Showcase Tip:
> **Why recruiters care:** Using `ChatPromptTemplate` to isolate your prompt engineering logic (system roles, rules, formatting constraints) from dynamic user inputs is a production-level standard. It demonstrates you write modular, clean AI architectures rather than messy concatenated strings.

### 🎯 Your Challenge:
1. **Define a Structured Template**: Create a `ChatPromptTemplate` with two distinct messages:
   - **System Message**: Set the persona of a stellar, conversion-focused B2B sales copywriter. Instruct it to keep the email under 150 words, avoid generic buzzwords ("innovative", "synergy", "revolutionary"), highlight a specific value proposition, and end with a clear, low-friction Call-To-Action (CTA).
   - **Human Message Template**: Use dynamic placeholders: `"Hello {prospect_name}, I noticed you lead operations at {company}..."` incorporating the prospect's background, your product, and the `{value_prop}`.
2. **Format and Inject CRM Data**: Define dynamic variables representing actual CRM values:
   * `prospect_name`: "Sarah Jenkins"
   * `company`: "Apex Logistics"
   * `value_prop`: "reduces delivery routing delays by 22% using predictive AI dispatch systems"
3. **Invoke and Output**: Run the formatted prompt through Gemini (`gemini-1.5-flash`) and output the outreach copy.

```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
import random

# Load environment variables
load_dotenv()

# Initialize Gemini model (temperature=0.5 fits creative yet professional copywriting)
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite", temperature=0.5)
```

```python
prompt_template = ChatPromptTemplate.from_messages([
    ("system", (
        "You are a professional B2B sales copywriter. "
        "Write outreach emails under 150 words. "
        "Avoid generic buzzwords like 'innovative', 'synergy', or 'revolutionary'. "
        "Highlight a specific value proposition and end with a clear, low-friction CTA."
    )),
    ("human", (
        "Write a cold outreach email to {prospect_name}, who leads operations at {company}. "
        "Our solution {value_prop}. Craft a compelling, concise email that gets a reply."
    ))
])
```

```python
# TODO: Format the template dynamically with CRM data
# 1. Define your inputs dictionary with 'prospect_name', 'company', and 'value_prop'
# 2. Call prompt_template.format_messages(**inputs)
# 3. Print the messages to verify the inputs are correctly mapped


person_name = ["Mike Sulley", "Abhis Khan", "Bill Walmart"]
company_name = ["Ant army logistics", "Cargo Sea Voyages", "Octadir delivery"]

selected_name = random.choice(person_name)
selected_company = random.choice(company_name)

inputs = {
  "prospect_name": selected_name,
  "company": selected_company,
  "value_prop": "reduces delivery routing delay by 22% using predictive AI"
}

messages = prompt_template.format_messages(**inputs)
print(messages)
```

```python
# TODO: Send the formatted messages list to the model and output the final email copy

# Your code here
response = llm.invoke(messages)
print(response.content)
```
