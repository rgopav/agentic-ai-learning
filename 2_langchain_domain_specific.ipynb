```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

from google.colab import userdata
api_key_value = userdata.get('GEMINI_API_KEY')

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    api_key=api_key_value,
    temperature=0
)

template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a US green energy infrastructure advisor. You only answer questions related to US EV infrastructure"),
        ("user", "{user}")
    ]
)

guardrail_chain = template | llm

# Fixed: changed sets to dictionaries using colons
out_of_context_question_1 = {"user": "Who is the president of United States"}
out_of_context_question_2 = {"user": "Where is the capital building of California located?"}

response1 = guardrail_chain.invoke(out_of_context_question_1)
response2 = guardrail_chain.invoke(out_of_context_question_2)
print("-" * 80)
print("\n")
print("Response 1:", response1.content)
print("-" * 80)
print("\n")

print("Response 2:", response2.content)
print("\n")
print("-" * 80)
```
